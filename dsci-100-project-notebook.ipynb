```r

#store the results in a new dataframe which I assigned to the object basketball_tidy_data_set, use pipeline to chain the subsequent operations
basketball_tidy_data_set <- womens_basketball |> 

#I want to use mutate() with tourney_finish to be more aligned with my question by creating a new column called lost_first_round 
#To have categorical outcome variable I will use factor() as it'll be TRUE or FALSE which I want to convert into categorical classes
# I use == to compare the values with first_round_loss in the tourney_finish column. This way R can check whether a teams tournament finish is first_round_loss 
#I added labels= c("No", "Yes") because I want FALSE to be changed to "No" and TRUE to be changed to "Yes"
mutate(lost_first_round = factor(tourney_finish == "first_round_loss",
                                   labels = c("No", "Yes"))) |>

#Interested in the variables seed and newly made column "lost_first_round" so extract them from the data set by using select function to choose a subset of a columns and eliminate the unnecessary columns
select(seed, reg_wins_pct, conf_wins_pct, total_wins_pct, lost_first_round)

#As the previous question, use head() to display the first six rows, as this function shows the first few rows and is set to show six
head(basketball_tidy_data_set) 
```

Visualization 1: The below plot is relevant to the analysis because as a seed number increases all the way from seed 1 to seed 16 we see a dramatic increase in the proportion of teams losing in the first round, this showcases a strong relationship between whether the team loses in the first round and their tournament seed. This direct separation shown clearly in this diagram suggests that the seed could be a useful predictor for the later KNN classification model.

```r
### EDIT THIS CODE CELL FOR VISUALIZATION 1###
#tournament seed versus losing in the first round.
options(repr.plot.width=9, repr.plot.height=6)
seed_graph <- basketball_tidy_data_set |>
filter(seed>0)|>
ggplot(aes(x=factor(seed), fill=lost_first_round))+
geom_bar(position = "fill")+
labs(x="The Tournament Seed", y="Proportion of Teams", fill= "Lost first round in tournament")+
ggtitle("Proportion of Womens Basketball Teams Losing in the First Round by Tournament Seed")
seed_graph
```

Visualization 2: The below plot is relevant to the analysis because it compares the two numerical variables tournament seed and regional win percentage directly with whether they lost in the first round of the tournament. We see a cluster of red dots signifying teams that did not lose in first round near the lower seed numbers and higher win percentages, whereas the blue dots signifying the team lost in the first round are found clustering in the top left with higher seeds and lower regional win percentages. This diagram demonstrates spatially that combining both of these predictors could let the KNN classifier separate the outcomes.

```r
### EDIT THIS CODE CELL FOR VISUALIZATION 2###
#regional win percentage versus losing in the first round.

reg_win_pct_graph <- basketball_tidy_data_set |>
filter(seed>0) |>
ggplot(aes(x=reg_wins_pct, y=seed, colour=lost_first_round))+
geom_point(alpha=0.4)+
labs(x="The regional win percentage(%)", y="The Tournament Seed", colour= "Lost first round in Tournament")+
ggtitle("Womens basketball regional win percentage versus losing first round in tournament")
reg_win_pct_graph
```
